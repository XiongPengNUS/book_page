# RSOME for Distributionally Robust Optimization

(rsome_dro:dro_framework)=
## The General Framework for Distributionally Robust Optimization Models

The RSOME package includes a `dro` module designed to facilitate modeling problems within a robust stochastic optimization framework, as proposed in {cite:ps}`Chen_Sim_Xiong_2020robust`. Built on top of the `ro` module, the dro module maintains consistency with the robust optimization modeling framework introduced in [RSOME for Robust Optimization](rsome_ro.ipynb), specifically in defining random variables and their associated uncertainty sets. Additionally, the `dro` module supports the definition of a class of event-wise ambiguity sets to capture distribution information, allowing the specification of objective functions or constraints involving worst-case expectations over the given ambiguity set. The general framework for such robust stochastic optimization models is presented below.

```{math}
\begin{aligned}
\min~&\sup\limits_{\mathbb{P}\in\mathcal{F}_0}\mathbb{E}_{\mathbb{P}}\left[\pmb{a}_0^{\top}(\tilde{s}, \tilde{\pmb{z}})\pmb{x}(\tilde{s}) + \pmb{b}_0^{\top}\pmb{y}(\tilde{s}, \tilde{\pmb{z}}) + c_0(\tilde{s}, \tilde{\pmb{z}})\right] \\
\text{s.t.}~&\pmb{a}_m^{\top}(s, \pmb{z})\pmb{x}(s) + \pmb{b}_m^{\top}\pmb{y}(s, \pmb{z}) + c_m(s, \pmb{z}) \leq 0 &&\forall \pmb{z}\in\mathcal{Z}_{m}, s\in[S], \forall m \in \mathcal{M}_1 \\
&\mathbb{E}_{\mathbb{P}}\left[\pmb{a}_m^{\top}(\tilde{s}, \tilde{\pmb{z}})\pmb{x}(\tilde{s}) + \pmb{b}_m^{\top}\pmb{y}(\tilde{s}, \tilde{\pmb{z}}) + c_m(\tilde{s}, \tilde{\pmb{z}})\right] \leq 0 &&\forall \mathbb{P}\in\mathcal{F}_m, \forall m \in \mathcal{M}_2 \\
&x_i \in \mathcal{A}\left(\mathcal{C}_x^i\right) &&\forall i \in [I_x] \\
&y_i \in \overline{\mathcal{A}}\left(\mathcal{C}_y^i, \mathcal{J}_y^i\right) &&\forall i \in [I_y] \\
&\pmb{x}(s)\in\mathcal{X} &&\forall s\in[S].
\end{aligned}
```

In this formulation, the parameters

```{math}
\begin{aligned}
&\pmb{a}_m(s, \pmb{z}) := \pmb{a}_{ms}^0 + \sum\limits_{j\in[J]}\pmb{a}_{ms}^jz_j \\
&\pmb{b}_m(s) := \pmb{b}_{ms} \\
&c_m(s, \pmb{z}) := c_{ms}^0 + \sum\limits_{j\in[J]}c_{ms}^jz_j
\end{aligned}
```

are defined for each scenario $s$, where $\pmb{a}_m(s, \pmb{z})$ and $c_m(s, \pmb{z})$ are affine functions of the random variable $\pmb{z}$. Constraints indexed by $m\in\mathcal{M}_1$ are enforced under all realizations of $\pmb{z}$ within the uncertainty set $\mathcal{Z}_{m}$. These constraints resemble the worst-case constraints detailed in the [RSOME for Robust Optimization](rsome_ro:ro_framework) section. Another set of constraints indexed by $m\in\mathcal{M}_2$ are imposed to consider the expectation across all possible distributions specified by an ambiguity set $\mathcal{F}_m$, following the general format introduced in {cite:ps}`Chen_Sim_Xiong_2020robust`:

```{math}
\begin{aligned}
\mathcal{F}_m = \left\{
\mathbb{P}\in\mathcal{P}_0\left(\mathbb{R}^J\times [S]\right) ~\left|~
\begin{array}{ll}
(\tilde{\pmb{z}}, \tilde{z}) \sim \mathbb{P} \\
\mathbb{E}_{\mathbb{P}}\left[\left.\tilde{\pmb{z}}\right|\tilde{s}\in \mathcal{E}_{km}\right]\in\mathcal{Q}_{km} & \forall k\in[K] \\
\mathbb{P}\left[\left.\tilde{\pmb{z}}\in\mathcal{Z}_{\tilde{s}m}\right|\tilde{s}=s\right] = 1 &\forall s\in [S] \\
\mathbb{P}[\tilde{s}=1] = p_s &\forall s \in [S] \\
\text{for some }\pmb{p} \in \mathcal{P}_m
\end{array}
\right.
\right\}
&& \forall m\in\mathcal{M}_2.
\end{aligned}
```

Here, for each constraint indexed by $m\in\mathcal{M}_2$,

- The conditional expectation of $\tilde{\pmb{z}}$ over events (defined as subsets of scenarios and denoted by $\mathcal{E}_{km}$) are known to reside in a conic representable set $\mathcal{Q}_{km}$;
- The support of $\tilde{\pmb{z}}$ in each scenario $s\in[S]$ is specified to be another conic representable set $\mathcal{Z}_{sm}$;
- Probabilities of scenarios, collectively denoted by a one-dimensional array $\pmb{p}$, are constrained by a third conic representable subset $\mathcal{P}_m\subseteq\left\{\left.\pmb{p}\in\mathcal{R}_{++}^S\right| \sum_{s\in[S]}=1\right\}$ in the probability simplex.

Dynamics of decision-making is captured by an *event-wise recourse adaptation* scheme for wait-and-see decisions proposed in {cite:ps}`Chen_Sim_Xiong_2020robust`: it allows the recourse decision to be adaptive to event realization, as in stochastic linear
optimization, and affinely adaptive to revealed uncertainty, as in distributionally robust optimization. When there is only one scenario (i.e., event), the event-wise recourse adaptation reduces to the affine recourse adaptation in {cite:ps}`Bertsimas_Sim_Zhang2019adaptive`. Given a collection $\mathcal{C}$ of mutually exclusive and collectively exhaustive events, an *event-wise static adaptation* scheme is defined for the decision $\pmb{x}(s)$ as

```{math}
\mathcal{A}(\mathcal{C}) := \left\{
x: [S]\mapsto \mathbb{R} ~\left|~
\begin{array}{l}
x(s) = x^{\mathcal{E}},~\mathcal{E}=\mathcal{H}_{\mathcal{H}}(s) \\
\text{for some }x^{\mathcal{E}}\in\mathbb{R}
\end{array}
\right.
\right\},
```

where $\mathcal{H}_{\mathcal{C}}:[S] \mapsto \mathcal{C}$ is a function such that $\mathcal{H}_{\mathcal{C}} = \mathcal{E}$ maps the scenario $s$ to the only event $\mathcal{E}$ in $\mathcal{C}$ that contains $s$. 

As for the recourse decision $\pmb{y}(s, \pmb{z})$, we define the *event-wise affine adaptation*:

```{math}
\overline{\mathcal{A}}(\mathcal{C}, \mathcal{J}) := \left\{
y: [S]\times\mathbb{R}^{J} ~\left|~
\begin{array}{l}
y(s, \pmb{z}) = y^0(s) + \sum\limits_{j\in\mathcal{J}}y^j(s)z_j \\
\text{for some }y^0, y^j\in\mathcal{A}(\mathcal{C}),~j\in\mathcal{J}
\end{array}
\right.
\right\},
```

for a subset $\mathcal{J}\subseteq [J]$ of random variables that the recourse decision affinely depends on. In the remaining part of the guide, we will introduce the RSOME code for specifying the event-wise ambiguity set and recourse adaptation rules.

## Introduction to the `rsome.dro` Environment

The `dro` module of the RSOME package maintains consistency with the previously introduced `ro` module. Consequently, most array operations, indexing, and slicing syntax applicable to `ro` models can be employed with `dro` models as well. The distinctive characteristics of the `dro` model primarily stem from its scenario-based representation of uncertainties and an alternative method for specifying the event-wise adaptation of decision variables.

Similar to `ro` models, `dro` model can be created by calling the `Model()` function imported from the `rsome.dro` module, as demonstrated by the code snippet below.

````python
from rsome import dro

model = dro.Model()
````

The `dro` models also support scenario-based uncertainties and adaptation schemes, with the number of scenarios specified through the `scens` argument when invoking the `Model()` function. Take the following code, for instance, a `dro` model with five scenarios is created.

````python
model = dro.Model(5)
````

(rsome_dro:rand_var)=
## Random Variables and Event-Wise Ambiguity Sets

### Random Variables
The syntax of creating random variables for a `dro` model is exactly the same as the `ro` models. Random variables are defined as arrays using the `rvar()` method of the `Model` object, where the input argument is used to specify the shape of the array, as demonstrated by the code below.

````python
from rsome import dro

model = dro.Model()

z = model.rvar()                            # a random variable as a scalar        
u = model.rvar(3)                           # 3 random variables as a 1D array
v = model.rvar((3, 5))                      # 3x5 random variables as a 2D array
w = model.rvar((2, 3, 4, 5))                # 2x3x4x5 random variables as a 4D array
````

Unlike `ro` models, the `dro` model is able to express the expectations of random variables, and this feature is utilized in specifying the uncertainty set $\mathcal{Q}_{km}$ within each ambiguity set $\mathcal{F}_m$. As demonstrated below, the expectations of random variables are indicated by the `E()` function imported from the RSOME package.

````python
from rsome import dro
from rsome import E
import rsome as rso

model = dro.Model()

u = model.rvar(3)                           # 3 random variables as a 1D array
v = model.rvar((3, 5))                      # 3x5 random variables as a 2D array

E(u) <= 1                                   # the expectation of u is smaller than or equal to one
E(v) >= 0                                   # the expectation of v is larger than or equal to zero
E(v).sum() == 1                             # sum of the expected value of v is one
rso.norm(E(u), 1) <= 1                      # one-norm of the expectation of u is no larger than one
````

### Event-Wise Ambiguity Sets

Ambiguity sets $\mathcal{F}_m$ of a `dro` model can be created by the `ambiguity()` method. 

````python
from rsome import dro

model = dro.Model(5)                        # create a model with 5 scenarios

fset = model.ambiguity()                    # create an ambiguity set of the model
print(fset)                                 # print scenarios of the model (ambiguity set)
````

RSOME supports indexing and slicing of the scenarios via either the labels or the integer-positions, as shown by the following code segments.

````python
print(fset[2])                              # the third scenario
````

````python
print(fset[:3])                             # the first three scenarios
````

The indexes of the scenarios are crucial in defining components of the ambiguity set, such as sets $\mathcal{Q}_{km}$ and $\mathcal{Z}_{sm}$, which will be discussed next.

### $\mathcal{Q}_{km}$ as the Uncertainty set of Conditional Expectations

According to the ambiguity set $\mathcal{Q}_{km}$ presented in the section of [The General Framework for Distributionally Robust Optimization Models](rsome_dro:dro_framework), the conic representable set $\mathcal{Q}_{km}$ is defined as the uncertainty set of the conditional expectation of random variables under the event $\mathcal{E}_k$ as a collection of selected scenarios. In the RSOME package, such a collection of scenarios can be specified by the indexing or slicing of the ambiguity set object, and constraints of the $\mathcal{Q}_{km}$ set are defined by the `exptset()` method of the ambiguity set object. Take the uncertainty sets below for the conditional expectation of $\tilde{\pmb{z}}$, for example,


```{math}
\begin{aligned}
&\mathbb{E}[\tilde{\pmb{z}}|s\in\mathcal{E}_1] \in \left\{
\pmb{z}: \mathbb{R}^3 \left|
\begin{array}{l}
|\pmb{z}| \leq 1\\
\|\pmb{z}\|_1 \leq 1.5
\end{array}
\right.
\right\}, &&\mathcal{E}_1 = \left\{0, 1, 2, 3, 4\right\} \\
&\mathbb{E}[\tilde{\pmb{z}}|s\in\mathcal{E}_2] \in \left\{
\pmb{z}: \mathbb{R}^3 \left|
\pmb{z} = 0
\right.
\right\}, &&\mathcal{E}_2 = \left\{0, 2, 4\right\}, \\
\end{aligned}
```

where the first event $\mathcal{E}_1$ is a collection of all five scenarios, and the second event $\mathcal{E}_2$ includes scenarios indexed by 0, 2, and 4. The conditional expectation information of the ambiguity set can be specified by the code below.

````python
from rsome import dro
from rsome import E
import rsome as rso

model = dro.Model(5)

z = model.rvar(3)

fset = model.ambiguity()                    # create an ambiguity set
fset.exptset(abs(E(z)) <= 1,
             rso.norm(E(z), 1) <= 1.5)      # the 1st uncertainty set of conditional expectations
fset[::2].exptset(E(z) == 0)                # the 2nd uncertainty set of conditional expectations
````

The ambiguity set `fset` itself represents the event $\mathcal{E}_1$ of all scenarios, and the `loc` indexer is used to form the event $\mathcal{E}_2$ with three selected scenarios. Besides `loc`, other indexing and slicing expressions described in the previous section can also be used to construct the events for the uncertainty sets of the expectations.

### $\mathcal{Z}_{sm}$ as the Scenario-Wise Support of Random Variables
The support $\mathcal{Z}_{sm}$ of random variables can be specified by the `suppset()` method, and the scenario information of the support can also be specified by the indexing and slicing expressions of the ambiguity set object. Take the following support of random variable $\tilde{\pmb{z}}\in\mathbb{R}^3$ for example,

```{math}
\begin{aligned}
&\mathbb{P}\left[
\left.
\tilde{\pmb{z}}\in \left\{
\pmb{z}: \mathbb{R}^3 \left|
\begin{array}{l}
|\pmb{z}| \leq 1\\
\|\pmb{z}\|_2 \leq 1.5
\end{array}
\right.
\right\}
\right| \tilde{s} = s
\right] = 1 &&\forall s \in \left\{0, 2, 4\right\} \\
&\mathbb{P}\left[
\left.
\tilde{\pmb{z}}\in \left\{
\pmb{z}: \mathbb{R}^3 \left|
\sum\limits_{j=1}^3z_j = 0
\right.
\right\}
\right| \tilde{s} = s
\right] = 1 &&\forall s \in \left\{1, 3\right\}, \\
\end{aligned}
```

the RSOME code for specifying the supports is provided below.

````python
from rsome import dro
from rsome import E
import rsome as rso

model = dro.Model(5)
z = model.rvar(3)

fset = model.ambiguity()
fset[::2].suppset(abs(z) <= 1,
                  rso.norm(z) <= 1.5)  # the support of z in scenarios 0, 2, 4
fset[1::2].suppset(z.sum() == 0)       # the support of z in scenarios 1, 3
````

Note that a valid ambiguity set must have the support sets for all scenarios to be specified. An error message will be given in solving the model if any of the supports are unspecified. RSOME provides a method called `showevents()` to display the specified supports for random variables and their expectations in a data frame, in order to help users check their ambiguity set.

````python
from rsome import dro
from rsome import E
from rsome import norm

model = dro.Model(5)
z = model.rvar(3)

fset = model.ambiguity()

fset[::2].suppset(abs(z) <= 1,
                       norm(z, 1) <= 1.5)
fset[1::2].suppset(z.sum() == 0)

fset.exptset(abs(E(z)) <= 1,
             rso.norm(E(z), 1) <= 1.5)
fset[::2].exptset(E(z) == 0)

fset.showevents()                           # display how the ambiguity set is specified
````

As the example above shows, supports for all scenarios have been defined, and there are two uncertainty sets defined for the conditional expectations.

### $\mathcal{P}_m$ as the Uncertainty Sets of Scenario Probabilities

In the event-wise ambiguity set, each scenario is assigned a probability $p_s$, and the values of the vector $\pmb{p}\in\mathbb{R}^S$ are characterized by an uncertainty set $\mathcal{P}_m$. In the RSOME package, this uncertainty set is defined using the `probset()` method of the ambiguity set object. For instance, the following code demonstrates the specification of an ellipsoidal uncertainty set for scenario probabilities.

````python
from rsome import dro
from rsome import norm

model = dro.Model(5)
z = model.rvar(3)

fset = model.ambiguity()
pr = model.p                                # p is the array of scenario probabilities
fset.probset(norm(pr - 0.2) <= 0.05)        # define the support of the array p
````

The scenario probabilities are formatted as a one-dimensional array, which can be accessed via the attribute `pr` of the model object. Notice that two underlying constraints for probabilities: $\pmb{p}\geq \pmb{0}$ and $\sum_{s=1}^Sp_s = 1$, are already integrated in the ambiguity set, so there is no need to specify them in defining the uncertainty set of $\pmb{p}$.

(rsome_dro:dec_var)=
## Decision Variables and Event-Wise Recourse Adaptations

Similar to RSOME ro models, decision variables in an RSOME `dro` model are also generated as arrays using the `dvar()` method. The shape of the array can be specified using the `shape` argument. As outlined in [The General Framework for Distributionally Robust Optimization Models](rsome_dro:dro_framework), decision variables can be event-wise static adaptive, adjusting to a set of events denoted by $\mathcal{C}$. Alternatively, they can be event-wise affinely adaptive, meaning that decisions adapt to events of $\mathcal{C}$ and also affinely depend on a subset of random variables indexed by $\mathcal{J}$. The event sets $\mathcal{C}$ and the subset of random variables can be specified using the `adapt()` method of decision variable objects, as demonstrated by the sample code below.

````python
from rsome import dro

model = dro.Model(5)
z = model.rvar(3)
x = model.dvar(3)

x.adapt(0)
x.adapt([2, 3])

x[:2].adapt(z[:2])
````

In the provided code, the set $\mathcal{C}$ is specified to include three events: $\{0\}$, $\{2, 3\}$, and collectively the remaining scenarios $\{1, 4\}$. Besides, the decision variable `x[:2]` is configured to be affinely adaptive to the random variable `z[:2]`. 

Note that decision variables created for RSOME `dro` models are initially non-adaptive, in the sense that the event set $\mathcal{C}=\{[S]\}$ and the dependent set $\mathcal{J} = \emptyset$. Therefore, if a decision variable is *here-and-now*, there is no need to specify the event set $\mathcal{C}$ and the subset $\mathcal{J}$ of random variable.

Similar to the linear decision rule defined for `ro` models, coefficients of an event-wise recourse adaptation could be accessed by the `get()` method after solving the model. Specifically, 

- `y.get()` returns the constant coefficients of the recourse adaptation. For models involving multiple scenarios, the returned object is a `pandas.Series` indexed by scenario labels, where each element of the series is an array that has the same shape as `y`.
- `y.get(z)` returns the linear coefficients of the recourse adaptation. For models of multiple scenarios, the returned object is a list. Each element of the list is an array, and the shape of the array is `y.shape + z.shape`, i.e., the combination of dimensions of `y` and `z`.

(rsome_dro:wc_expectations)=
## The Worst-Case Expectations

Once the ambiguity sets $\mathcal{F}_m$ and the recourse adaptation of decision variables are defined, expressions involving expectations under the worst-case distribution within the $\mathcal{F}$ set can be specified. If the objective function involves worst-case expectations, the RSOME `dro` model provides the `minsup()` method to minimize the supremum of the expectation term. Similarly, the `maxinf()` method can be utilized to maximize the infimum of the expectation term. Take the following distributionally robust model for example,

```{math}
\begin{aligned}
\max~&\sup\limits_{\mathbb{P}\in\mathcal{F}}\mathbb{E}_{\mathbb{P}}\left[\tilde{\pmb{z}}^{\top}\pmb{x}\right] \\
\text{s.t. }&\pmb{x} = \pmb{1},
\end{aligned}
```

with the ambiguity set $\mathcal{F}$ written as

```{math}
\mathcal{F} = \left\{
\mathbb{P} \in \mathcal{P}_0\left(\mathbb{R}^I\times\mathbb{R}\times[S]\right) \left|
\begin{array}{ll}
(\tilde{\pmb{z}}, \tilde{u}, \tilde{s}) \sim \mathbb{P} \\
\mathbb{E}_{\mathbb{P}}[u] \leq 0.01 \\
\mathbb{P}[\|\tilde{\pmb{z}} - \hat{\pmb{z}}_s\|_1 \leq u|\tilde{s}=s]=1 &\forall s \in [S] \\
\mathbb{P}[\tilde{s}=s] = 1/S &\forall s\in[S]
\end{array}
\right.
\right\},
```

where $\hat{\pmb{z}}_s$ is the $s$th record of a sample representing the distribution of the random variable $\tilde{\pmb{z}}$. Such a model can be solved using the code below.

````python
from rsome import dro
from rsome import E
import rsome as rso
import numpy as np

S, I = 10, 5
np.random.seed(0)
zhat = np.random.rand(S, I)

model = dro.Model(S)
z = model.rvar(I)
u = model.rvar()

fset = model.ambiguity()
for s in range(S):
    fset[s].suppset(rso.norm(z - zhat[s], 1) <= u)
fset.exptset(E(u) <= 0.01)
pr = model.p
fset.probset(pr == 1/S)

x = model.dvar(I)

model.maxinf(E(z@x), forall=fset)
model.st(x == 1)

model.solve()
````

Here, the objective is to maximize the expectation term `E(z@x)` under the worst-case distribution within the ambiguity set `fset` specified as the second argument of the `maxinf()` method of the `dro` model.

Similar to `ro` models, ambiguity sets of constraints indexed by $m\in\mathcal{M}_2$, concerning the worst-case expectations, can be specified by the `forall()` method, and if the ambiguity set of such constraints are unspecified, then by default, the underlying ambiguity set is the same as $\mathcal{F}_0$ defined for the objective function. As for the worst-case constraints indexed by $\mathcal{M}_1$, the uncertainty set can also be specified by the `forall()` method, and if the uncertainty set is unspecified, such worst-case constraints are defined upon support sets $\mathcal{Z}_{0s}$ of the default ambiguity set $\mathcal{F}_0$, for each scenario $s\in[S]$.
