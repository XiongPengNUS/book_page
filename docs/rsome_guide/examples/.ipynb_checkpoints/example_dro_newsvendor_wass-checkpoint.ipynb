# Robust Newsvendor under Wasserstein Ambiguity

In this example, we use RSOME to solve the multi-item newsvendor problem discussed in {cite:ps}`Chen_Sim_Xiong_2020robust`. This newsvendor problem determines the order quantity $w_i$ of each product $i$ under an overall budget of $d$. There are totally $I$ products, and each product has a selling price $p_i$ and an ordering cost $c_i$. The uncertain demands of these products are represented by the random variable $\tilde{\pmb{z}}\in\mathbb{R}^I$. Once the demand quantities are realized, the selling quantity $y_i$ of each product is expressed as $\min\left\{x_i, z_i\right\}$, and the newsvendor problem can be formulated as the following distributionally robust optimization model:

$$
\begin{align}
\min~&-\pmb{p}^{\top}\pmb{x} + \sup\limits_{\mathbb{P}\in\mathcal{F}_W(\theta)} \mathbb{E}_{\mathbb{P}}\left[
\sum\limits_{i\in[I]}p_i\max\{x_i - \tilde{z}_i, 0\}
\right] \\
\text{s.t. }&\pmb{c}^{\top}\pmb{x} = d \\
&\pmb{x}\in\mathbb{R}_+^I,
\end{align}
$$

where $\mathcal{F}_W(\theta)$ is an ambiguity set that captures all distributions of $\tilde{\pmb{z}}$ within a Wasserstein ball of radius $\theta \geq 0$ centered at the empirical distribution. Such an ambiguity set is expressed as the following even-wise format:

$$
\mathcal{F}_W(\theta) = \left\{
\mathbb{P}\in\mathcal{P}_0\left(\mathbb{R}^I\times\mathbb{R}\times [S]\right) 
\left|
\begin{array}{ll}
(\tilde{\pmb{z}}, \tilde{u}, \tilde{s}) \in \mathbb{P} \\
\mathbb{E}_{\mathbb{P}}[\tilde{u}|\tilde{s}\in[S]] \leq \theta \\
\mathbb{P}\left[\left.(\tilde{\pmb{z}}, \tilde{u})\in\mathcal{Z}_s \right| \tilde{s}=s\right] = 1 &\forall s \in [S] \\
\mathbb{P}[\tilde{s}=s] = \frac{1}{S}
\end{array}
\right.
\right\},
$$

where $S$ is the sample size of the dataset representing the empirical distribution, and the support $\mathcal{Z}$ is defined for each sample record $s$ as:

$$
\mathcal{Z}_s = \left\{
(\pmb{z}, u): \pmb{0} \leq \pmb{z}\leq \bar{\pmb{z}}, ~\|\pmb{z} - \hat{\pmb{z}}_s\|_1 \leq u
\right\}.
$$

Notice that besides the random variable $\tilde{\pmb{z}}$, an auxiliary variable $\tilde{u}$ is introduced to express the epigraph of one-norm distance $\|\pmb{z} - \hat{\pmb{z}}_s\|_1$.

In the subsequent numerical experiments, parameters of the model and ambiguity set are specified as follows:
- The number of products is $I=6$;
- The unit cost of each product is $c_i = 1$;
- The unit price of each product $p_i$ is randomly generated from a uniform distribution within $[1, 5]$.
- The budget of product ordering is $d=50I$. 
- The upper bound of each product demand $\bar{z}_i$ is randomly generated from the uniform distribution on $[0, 100]$;
- The empirical data has a sample size of $S=50$;
- The empirical distribution of demands is represented by the array $\hat{\pmb{z}}\in\mathbb{R}^{S\times I}$ randomly generated from the uniform distribution on $[\pmb{0}, \bar{\pmb{z}}]$;
- The Wasserstein metric parameter $\theta=0.01 \min \{\bar{\pmb{z}}\}$.

````python
import numpy as np

np.random.seed(1)

I = 6
c = np.ones(I)
p = 1 + 4*np.random.rand(I)
d = 50 * I
zbar = 100 * np.random.rand(I)

S = 50
zhat = zbar * np.random.rand(S, I)
theta = 0.01 * zbar.min()
````

In the objective function of the robust model, $\sum_{i\in[I]}p_i\max\{x_i - \tilde{z}_i, 0\}=\max_{\mathcal{J}\subseteq [I]}\sum_{j\in\mathcal{J}}p_i(x_i - \tilde{z}_i)$ is convex and piecewise affine involving $2^I$ pieces. Such a piecewise expression can be defined using the `maxof()` function imported from the RSOME package, as demonstrated by the code below.

````python
from itertools import product
from rsome import dro
from rsome import E
import rsome as rso

model = dro.Model(S)
z = model.rvar(I)
u = model.rvar() 

fset = model.ambiguity()
for s in range(S):
    fset[s].suppset(0 <= z, z <= zbar,
                    rso.norm(z - zhat[s], 1) <= u)
fset.exptset(E(u) <= theta)
pr = model.p 
fset.probset(pr == 1/S)

x = model.dvar(I)

pieces = [np.array(b)@(p*(x - z)) 
          for b in product([0, 1], repeat=I)]

model.minsup(-p@x + E(rso.maxof(pieces)), fset)
model.st(c@x == d, x >= 0)

model.solve()
````

The objective value in terms of the worst-case expected total cost can be retrieved as follows.

````python
print(f'Objective: {model.get():.5f}')
````

The distributionally robust model is a linear programming problem with dimension information provided below.

````python
model.do_math()
````

It can be seen that the linear programming problem has a large dimension, because the number of pieces involved in the piecewise affine objective function grows exponentially as the product number increases. We will then demonstrate the approaches mentioned in {cite:ps}`Chen_Sim_Xiong_2020robust` that applies event-wise affine  adaptation to approximate the piecewise affine function. Here, a wait-and-see decision $y_i$ is introduced to represent the piecewise expression $\max\{x_i - z_i, 0\}$, and the recourse decision is further approximated by the adaptation scheme $\overline{\mathcal{A}}(\{\{1\}, \{2\}, ..., \{S\}\}, [I+1])$, which implies that the decision adapts to each scenario $s$ and also affinely dependents on random variables $\pmb{z}\in\mathbb{R}^I$ and $u\in\mathbb{R}$. Such an adaptive distributionally robust model is implemented by the following code.

````python
from rsome import dro
from rsome import E
import rsome as rso

model = dro.Model(S)
z = model.rvar(I)
u = model.rvar() 

fset = model.ambiguity()
for s in range(S):
    fset[s].suppset(0 <= z, z <= zbar,
                    rso.norm(z - zhat[s], 1) <= u)
fset.exptset(E(u) <= theta)
pr = model.p 
fset.probset(pr == 1/S)

x = model.dvar(I)
y = model.dvar(I)
y.adapt(z) 
y.adapt(u)
for s in range(S):
    y.adapt(s)

model.minsup(-p@x + E(p@y), fset)
model.st(y >= x - z, y >= 0)
model.st(c@x == d, x >= 0)

model.solve()
````

It can be seen that by utilizing recourse decision $\pmb{y}$ following the event-wise affine adaptation, the same objective value is achieved as below, with much shorter solution time. This is because the dimension of the problem is much smaller compared with the previous model. 

````python
print(f'Objective: {model.get():.5f}')
````

````python
model.do_math()
````
