## Installation via PyPI

The RSOME package can be installed with the `pip` command:

```console
pip install rsome
```

regardless of the versions of Python or the operating systems.

The following linear optimization problem
```{math}
\begin{aligned}
\max~&0.2x_1 + 0.4x_2 + 0.3x_3 \\
\text{s.t.}~&\|\pmb{x}\|_2 \leq 1
\end{aligned}
```
can be implemented as follows to test the installation of the RSOME package.

````python
import numpy as np
from rsome import ro                # import the ro modeling tool

c = np.array([0.2, 0.4, 0.3])       # coefficients of the objective function

model = ro.Model()                  # create a Model object
x = model.dvar(3)                   # define x as an array of three variables

model.max(c @ x)                    # maximize the objective function
model.st(x.norm() <= 1)             # specify the constraint

model.solve()                       # solve the model by the default solver
````

If the installation is successful, the problem will be solved by the default `scipy.optimize` solver (see the documentation [here](https://docs.scipy.org/doc/scipy/reference/optimize.html)), and the solution status and running time will be printed. We can then access the optimal solution and the corresponding objective value can be attained by the `get()` method.

````python
print(f'x: {x.get()}')
print(f'Objective: {model.get()}'.format())
````

## Dependencies

The minimum dependencies of the current version of RSOME are:

- [Python](https://www.python.org/) >= 3.9
- [Pandas](https://pandas.pydata.org/) >= 0.25.0
- [NumPy](https://numpy.org/) >= 1.26.4
- [Numba](https://numba.pydata.org) >= 0.65.1
- [SciPy](https://scipy.org/) >= 1.17.1
- [ECOS](https://github.com/embotech/ecos) >= 2.0.10
- [Clarabel](https://clarabel.org/stable/) >= 0.11.0

The installation of solvers is optional and the details can be found below.


(label-solver-installation)=
## Installation of Solvers

In addition to the default linear programming solver `scipy.optimize`, RSOME offers modeling tools and interfaces for a range of open-source and commercial solvers. However, integrating alternative solvers requires additional installation steps. For comprehensive instructions on installing these solvers, please refer to the following table, which directs you to their required versions respective websites.

| Solver | License | Name | RSOME Interface | Website
|:-------|:-----------|:------|:----------------|:--------------|
|SciPy >= 1.9.0 | Open-source | `'scipy'` | `lpg_solver` | https://docs.scipy.org/doc/scipy/reference/optimize.html |
|ECOS >= 2.0.10 | Open-source | `'ecos'` | `eco_solver` | https://github.com/embotech/ecos-python |
|Clarabel >= 0.11.1 | Open-source | `'clarabel'` | `clb_solver` | https://github.com/oxfordcontrol/Clarabel.jl|
|Gurobi >= 11.0.0 | Commercial | `'gurobi'` | `grb_solver` | https://www.gurobi.com/ |
|Mosek >= 10.0.44 | Commercial | `'mosek'` | `msk_solver` | https://www.mosek.com/ |
|COPT >= 7.0.3 | Commercial | `'copt'` | `cpt_solver` | https://www.shanshu.ai/copt |

Consider the commercial COPT solver as an example; once successfully installed, it can be applied to solve the RSOME model via the imported solver interface object or the solver name `'copt'`.

````python
from rsome import cpt_solver as cpt

model.solve(cpt)             # Use the solver interface cpt_solver
````

````python
model.solve('copt')          # Use the solver name 'copt'
````
