# Box with the Maximum Volume

A manufacturer wants a rectangular box with maximum volume subject to material limits and proportionality constraints. The following model is thus used to optimize the shape of the box, in terms of its width $w$, height $h$, and depth $d$, so that its volume is maximized:

```{math}
\begin{aligned}
\max~& whd \\
\text{s.t.}~&2(wh + dh) \leq A_{\text{wall}} \\
&wd \leq A_{\text{floor}} \\
&\alpha \leq h/w \leq \beta, \gamma \leq d/w \leq \delta, 
\end{aligned}
```

where model parameters:
- Limits on wall area $A_{\text{wall}} = 200$,
- Limits on floor area $A_{\text{floor}} = 150$,
- Limits on height-width ratio: $\alpha=0.8$, and $\beta=1.5$,
- Limits on depth-width ratio: $\gamma=0.8$, and $\delta=1.5$.

A logarithmic change of variables turns the multiplicative geometry into a convex exponential-cone model: $x = \log(w)$, $y = \log(h)$, and $z = \log(d)$, then we have

```{math}
\begin{aligned}
\max~& x + y + z \\
\text{s.t.}~&2(\exp(x + y) + \exp(z + y)) \leq A_{\text{wall}} \\
&\exp(x + z) \leq A_{\text{floor}} \\
&\exp(x - y) \leq \alpha, \exp(y - x) \leq \beta \\
&\exp(x - z) \leq \gamma, \exp(z - x) \leq \delta. \\
\end{aligned}
```

The conic model can be implemented and solved by the code below.

````python
from rsome import ro
import rsome as rso
import numpy as np

A_wall = 200
A_floor = 150
alpha, beta = 0.8, 1.5
gamma, delta = 0.8, 1.5

model = ro.Model()
x = model.dvar()
y = model.dvar()
z = model.dvar()

model.max(x + y + z)
model.st(2*(rso.exp(x + y) + rso.exp(z + y)) <= A_wall)
model.st(rso.exp(x + z) <= A_floor)
model.st(rso.exp(x - y) <= alpha, rso.exp(y - x) <= beta)
model.st(rso.exp(x - z) <= gamma, rso.exp(z - x) <= delta)
model.solve()
````

The optimal solution is presented in the following code segment.

````python
print(x.get(), y.get(), z.get())
````

The dimension of the box is retrieved by the following code.

````python
print(f'Width:  {np.exp(x.get()):.4f}')
print(f'Height: {np.exp(y.get()):.4f}')
print(f'Depth:  {np.exp(z.get()):.4f}')
````
